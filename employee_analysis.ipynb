## 6. Data Visualization

```python
sns.countplot(data=df, x="Education", hue="LeaveOrNot")
plt.title("Employee Leave by Education")
plt.xlabel("Education")
plt.ylabel("Number of Employees")
plt.show()
```

```python
sns.countplot(data=df, x="City", hue="LeaveOrNot")
plt.title("Employee Leave by City")
plt.xlabel("City")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.countplot(data=df, x="Age", hue="LeaveOrNot")
plt.title("Employee Leave by Age")
plt.xlabel("Age")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.countplot(data=df, x="Gender", hue="LeaveOrNot")
plt.title("Employee Leave by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Employees")
plt.show()
```

```python
sns.countplot(data=df, x="ExperienceInCurrentDomain", hue="LeaveOrNot")
plt.title("Employee Leave by Experience")
plt.xlabel("Experience in Current Domain (Years)")
plt.ylabel("Number of Employees")
plt.show()
```

```python
sns.countplot(data=df, x="PaymentTier", hue="LeaveOrNot")
plt.title("Employee Leave by Payment Tier")
plt.xlabel("Payment Tier")
plt.ylabel("Number of Employees")
plt.show()
```

```python
sns.countplot(data=df, x="EverBenched", hue="LeaveOrNot")
plt.title("Employee Leave vs Bench Status")
plt.xlabel("Ever Benched")
plt.ylabel("Number of Employees")
plt.show()
```

## 8. Leave Rate Analysis

```python
leave_by_education = (
    df.groupby("Education")["LeaveOrNot"]
    .mean()
    .mul(100)
    .round(2)
)

leave_by_education
```

```python
leave_by_gender = (
    df.groupby("Gender")["LeaveOrNot"]
    .mean()
    .mul(100)
    .round(2)
)

leave_by_gender
```

```python
leave_by_payment = (
    df.groupby("PaymentTier")["LeaveOrNot"]
    .mean()
    .mul(100)
    .round(2)
)

leave_by_payment
```

```python
leave_by_city = (
    df.groupby("City")["LeaveOrNot"]
    .mean()
    .mul(100)
    .round(2)
    .sort_values(ascending=False)
)

leave_by_city
```

```python
leave_by_experience = (
    df.groupby("ExperienceInCurrentDomain")["LeaveOrNot"]
    .mean()
    .mul(100)
    .round(2)
)

leave_by_experience
```

```python
overall_leave_rate = df["LeaveOrNot"].mean() * 100

print(f"Overall employee leave rate: {overall_leave_rate:.2f}%")
```

## 9. Key Findings

```python
print("Highest leave rate by city:")
print(
    df.groupby("City")["LeaveOrNot"]
    .mean()
    .mul(100)
    .round(2)
    .sort_values(ascending=False)
    .head(3)
)

print("\nHighest leave rate by experience:")
print(
    df.groupby("ExperienceInCurrentDomain")["LeaveOrNot"]
    .mean()
    .mul(100)
    .round(2)
    .sort_values(ascending=False)
    .head(3)
)
```
